```python
data = "42 3.14 hello".split()
i = int(data[0])
f = float(data[1])
s = str(data[2])
print(f"Integer value: {i}, type: int")
print(f"Float value: {round(f,1)}, type: float")
print(f"String value: {s}, type: str")
```
